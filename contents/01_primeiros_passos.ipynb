## 1. Expressões e variáveis

Altere os valores e execute a célula novamente.

```python
nome = "Ana"
nota_1 = 8.0
nota_2 = 9.5
media = (nota_1 + nota_2) / 2

print(f"Olá, {nome}!")
print(f"Sua média é {media:.1f}.")
```

## 2. Condicionais

```python
temperatura = 27

if temperatura >= 30:
    print("Está quente.")
elif temperatura >= 20:
    print("A temperatura está agradável.")
else:
    print("Está frio.")
```

## 3. Repetição

```python
for numero in range(1, 6):
    print(numero, numero ** 2)
```

## 4. Funções

Uma função permite encapsular uma tarefa e reutilizá-la.

```python
def quadrado(x):
    return x * x

for x in range(1, 6):
    print(x, quadrado(x))
```
